```python
# Remove null job titles
df = df.dropna(subset=['job_title']).copy()

# Fill missing values
df['seniority_level'] = df['seniority_level'].fillna('Not Specified')

df['status'] = df['status'].fillna('Unknown')

df['location'] = df['location'].fillna('Unknown')

df['ownership'] = df['ownership'].fillna('Unknown')

df['revenue'] = df['revenue'].fillna('Not Mentioned')

# Create days_ago column
df['days_ago'] = df['post_date'].str.extract(r'(\d+)')

# Convert to numeric
df['days_ago'] = pd.to_numeric(
    df['days_ago'],
    errors='coerce'
)

# Fill missing values
df['days_ago'] = df['days_ago'].fillna(0)

# Convert to integer
df['days_ago'] = df['days_ago'].astype(int)

df.info()
```

```python
df.to_csv(
    "../cleaned_data/cleaned_job_market_2025.csv",
    index=False
)
df.head()
df.info()
```

# KPI Analysis

# BIVARIATE ANALYSIS
(Two variables analysis)

```python
import re
import pandas as pd

def extract_salary(value):
    value = str(value)
    value = value.replace(',', '')
    
    numbers = re.findall(r'\d+', value)
    
    if len(numbers) >= 2:
        return (int(numbers[0]) + int(numbers[1])) / 2
    elif len(numbers) == 1:
        return int(numbers[0])
    else:
        return None

df['salary_numeric'] = df['salary'].apply(extract_salary)
```

```python
# Average Salary by Seniority
df.groupby('seniority_level')['salary_numeric'].mean().sort_values(ascending=False)
```

```python
# Jobs by Industry
df.groupby('industry')['job_title'].count().sort_values(ascending=False).head(10)
```

```python
# Company Size vs Job Count
df.groupby('company_size')['job_title'].count().sort_values(ascending=False)
```

```python
# Location vs Jobs
df.groupby('location')['job_title'].count().sort_values(ascending=False).head(10)
```

```python
# Top Job Roles

plt.figure(figsize=(10,6))

top_roles.plot(kind='bar')

plt.title("Top 10 Job Roles")

plt.xlabel("Job Role")
plt.ylabel("Count")

plt.xticks(rotation=45)

plt.savefig("../charts/top_job_roles.png")

plt.show()
```

```python
# Top Skills Chart

top_skills = df['skills'].str.split(',').explode().str.strip().value_counts().head(10)

plt.figure(figsize=(10,6))
top_skills.sort_values().plot(kind='barh')
plt.title("Top 10 Most Demanded Skills")
plt.xlabel("Job Count")
plt.ylabel("Skills")
plt.savefig("../charts/top_skills.png", bbox_inches="tight")
plt.show()
```

```python
df['days_ago'] = df['post_date'].str.extract(r'(\d+)')
df['days_ago'] = pd.to_numeric(df['days_ago'], errors='coerce')
df.head()
```

```python
#  Job Posting Trend

jobs_by_day = df['days_ago'].value_counts().sort_index()

plt.figure(figsize=(10,5))

plt.plot(
    jobs_by_day.index,
    jobs_by_day.values,
    marker='o'
)

plt.title("Job Posting Trend by Days Ago")
plt.xlabel("Days Ago")
plt.ylabel("Number of Jobs")

plt.savefig("../charts/job_posting_trend.png", bbox_inches="tight")

plt.show()
```

```python
# Seniority Level

seniority_count = df['seniority_level'].value_counts()

plt.figure(figsize=(7,7))
plt.pie(
    seniority_count,
    labels=seniority_count.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title("Seniority Level Distribution")
plt.savefig("../charts/seniority_distribution.png", bbox_inches="tight")
plt.show()
```

```python
# Top Locations

top_locations = df['location'].value_counts().head(10)

plt.figure(figsize=(10,6))
top_locations.plot(kind='bar')
plt.title("Top 10 Hiring Locations")
plt.xlabel("Location")
plt.ylabel("Job Count")
plt.xticks(rotation=45, ha='right')
plt.savefig("../charts/top_locations.png", bbox_inches="tight")
plt.show()
```

```python
# Top Companies

top_companies = df['company'].value_counts().head(10)

plt.figure(figsize=(10,6))
top_companies.sort_values().plot(kind='barh')
plt.title("Top 10 Hiring Companies")
plt.xlabel("Job Count")
plt.ylabel("Company")
plt.savefig("../charts/top_companies.png", bbox_inches="tight")
plt.show()
```

```python
# Industry Distribution

industry_count = df['industry'].value_counts()

top5 = industry_count.head(5)

others = pd.Series(
    {'Others': industry_count.iloc[5:].sum()}
)

final_industry = pd.concat([top5, others])

plt.figure(figsize=(8,8))

plt.pie(
    final_industry,
    labels=final_industry.index,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'width': 0.4}
)

plt.title("Top Industry Distribution")

plt.savefig(
    "../charts/industry_distribution.png",
    bbox_inches="tight"
)

plt.show()
```

```python
# Salary Distribution

import re

def extract_salary(value):
    nums = re.findall(r'\d+', str(value).replace(',', ''))
    if len(nums) >= 2:
        return (int(nums[0]) + int(nums[1])) / 2
    elif len(nums) == 1:
        return int(nums[0])
    else:
        return None

df['salary_numeric'] = df['salary'].apply(extract_salary)

plt.figure(figsize=(10,5))
plt.hist(df['salary_numeric'].dropna(), bins=20)
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Job Count")
plt.savefig("../charts/salary_distribution.png", bbox_inches="tight")
plt.show()
```
